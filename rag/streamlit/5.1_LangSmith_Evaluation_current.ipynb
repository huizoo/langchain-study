# 3. Retriever 생성

- Evaluation을 위한 Retriever 생성
- 3.x 에서 생성한 Pinecone Retriever 

```python
from langchain_openai import OpenAIEmbeddings

embedding = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
from langchain_pinecone import PineconeVectorStore

index_name = 'tax-markdown-index'

database = PineconeVectorStore.from_existing_index(index_name=index_name, embedding=embedding)
retriever = database.as_retriever()
```

# 4. LLM 답변 생성을 위한 RagBot 

- retriever와 OpenAI API를 활용한 RAG

```python
import os
from langchain_openai import ChatOpenAI
from langsmith import traceable

llm = ChatOpenAI(
    model=os.getenv("RAG_MODEL", "gpt-5-nano"),
    temperature=0)

@traceable()
def rag_bot(question: str) -> dict:
    docs = retriever.invoke(question)

    docs_string = "\n\n".join(doc.page_content for doc in docs)

    instructions = f"""
    당신은 한국의 소득세 전문가입니다.
    아래 소득세법을 참고해서 사용자의 질문에 답변해주세요.

    규칙:
    - 문서에 근거해서 답변하세요.
    - 문서에서 알 수 없으면 모른다고 답변하세요.
    - 답변은 간결하게 작성하세요.

    Documents:
    {docs_string}""".strip()

    ai_msg = llm.invoke([
            {"role": "system", "content": instructions},
            {"role": "user", "content": question},
        ],
    )
    
    return {"answer": ai_msg.content, "documents": docs}
```

```python
from typing_extensions import Annotated, TypedDict

judge_llm = ChatOpenAI(
    model=os.getenv("JUDGE_MODEL", "gpt-5-nano"),
    temperature=0,
)
```

```python
class CorrectnessGrade(TypedDict):
    explanation: Annotated[str, ..., "평가 근거"]
    correct: Annotated[bool, ..., "정답과 의미상 일치하면 True, 아니면 False"]


correctness_instructions = """
당신은 한국어 QA 평가자입니다.
QUESTION, GROUND TRUTH ANSWER, STUDENT ANSWER가 주어집니다.

평가 기준:
1. STUDENT ANSWER가 GROUND TRUTH ANSWER와 의미상 일치하는지 평가하세요.
2. 표현이 달라도 핵심 의미가 같으면 correct=True입니다.
3. 핵심 사실이 틀렸거나 빠졌거나 충돌하면 correct=False입니다.
4. STUDENT ANSWER에 추가 정보가 있어도 정답과 충돌하지 않으면 허용합니다.

판단 근거를 먼저 설명한 뒤 correct 값을 반환하세요.
""".strip()


correctness_llm = judge_llm.with_structured_output(
    CorrectnessGrade,
    method="json_schema",
    strict=True,
)


def correctness(inputs: dict, outputs: dict, reference_outputs: dict) -> bool:
    answer = f"""
QUESTION: {inputs["input_question"]}

GROUND TRUTH ANSWER: {reference_outputs["output_answer"]}

STUDENT ANSWER: {outputs["answer"]}
""".strip()

    grade = correctness_llm.invoke(
        [
            {"role": "system", "content": correctness_instructions},
            {"role": "user", "content": answer},
        ]
    )

    return grade["correct"]
```

```python
class RelevanceGrade(TypedDict):
    explanation: Annotated[str, ..., "평가 근거"]
    relevant: Annotated[bool, ..., "질문에 직접적으로 답하면 True, 아니면 False"]


relevance_instructions = """
당신은 한국어 QA 평가자입니다.
QUESTION과 STUDENT ANSWER가 주어집니다.

평가 기준:
1. STUDENT ANSWER가 QUESTION에 직접적으로 답하는지 평가하세요.
2. 답변이 질문 해결에 도움이 되면 relevant=True입니다.
3. 질문과 무관하거나 핵심을 피하면 relevant=False입니다.

판단 근거를 먼저 설명한 뒤 relevant 값을 반환하세요.
""".strip()


relevance_llm = judge_llm.with_structured_output(
    RelevanceGrade,
    method="json_schema",
    strict=True,
)


def relevance(inputs: dict, outputs: dict) -> bool:
    answer = f"""
QUESTION: {inputs["input_question"]}

STUDENT ANSWER: {outputs["answer"]}
""".strip()

    grade = relevance_llm.invoke(
        [
            {"role": "system", "content": relevance_instructions},
            {"role": "user", "content": answer},
        ]
    )

    return grade["relevant"]
```

```python
class GroundedGrade(TypedDict):
    explanation: Annotated[str, ..., "평가 근거"]
    grounded: Annotated[bool, ..., "답변이 검색 문서에 근거하면 True, 아니면 False"]


grounded_instructions = """
당신은 RAG 답변의 근거성을 평가하는 평가자입니다.
FACTS와 STUDENT ANSWER가 주어집니다.

평가 기준:
1. STUDENT ANSWER의 핵심 주장이 FACTS에 의해 뒷받침되는지 평가하세요.
2. FACTS에 없는 내용을 단정하거나 FACTS와 충돌하면 grounded=False입니다.
3. FACTS에 근거해 답변했다면 grounded=True입니다.

판단 근거를 먼저 설명한 뒤 grounded 값을 반환하세요.
""".strip()


grounded_llm = judge_llm.with_structured_output(
    GroundedGrade,
    method="json_schema",
    strict=True,
)


def groundedness(inputs: dict, outputs: dict) -> bool:
    doc_string = "\n\n".join(
        doc.page_content for doc in outputs["documents"]
    )

    answer = f"""
FACTS:
{doc_string}

STUDENT ANSWER:
{outputs["answer"]}
""".strip()

    grade = grounded_llm.invoke(
        [
            {"role": "system", "content": grounded_instructions},
            {"role": "user", "content": answer},
        ]
    )

    return grade["grounded"]
```

```python
class RetrievalRelevanceGrade(TypedDict):
    explanation: Annotated[str, ..., "평가 근거"]
    relevant: Annotated[bool, ..., "검색 문서가 질문과 관련 있으면 True, 아니면 False"]


retrieval_relevance_instructions = """
당신은 RAG 검색 결과의 관련성을 평가하는 평가자입니다.
QUESTION과 FACTS가 주어집니다.

평가 기준:
1. FACTS가 QUESTION에 답변하는 데 관련 있는지 평가하세요.
2. 일부 불필요한 내용이 섞여 있어도 핵심적으로 관련 있으면 relevant=True입니다.
3. FACTS가 QUESTION과 완전히 무관하면 relevant=False입니다.

판단 근거를 먼저 설명한 뒤 relevant 값을 반환하세요.
""".strip()


retrieval_relevance_llm = judge_llm.with_structured_output(
    RetrievalRelevanceGrade,
    method="json_schema",
    strict=True,
)


def retrieval_relevance(inputs: dict, outputs: dict) -> bool:
    doc_string = "\n\n".join(
        doc.page_content for doc in outputs["documents"]
    )

    answer = f"""
QUESTION:
{inputs["input_question"]}

FACTS:
{doc_string}
""".strip()

    grade = retrieval_relevance_llm.invoke(
        [
            {"role": "system", "content": retrieval_relevance_instructions},
            {"role": "user", "content": answer},
        ]
    )

    return grade["relevant"]
```

```python
def target(inputs: dict) -> dict:
    return rag_bot(inputs["input_question"])
```

```python
experiment_results = client.evaluate(
    target,
    data=dataset_name,
    evaluators=[
        correctness,
        relevance,
        groundedness,
        retrieval_relevance,
    ],
    experiment_prefix="income-tax-rag-function-eval",
    metadata={
        "version": "function-style-rag",
        "rag_model": os.getenv("RAG_MODEL", "gpt-5-nano"),
        "judge_model": os.getenv("JUDGE_MODEL", "gpt-5-nano"),
        "retriever": "pinecone-tax-markdown-index",
    },
)
```
